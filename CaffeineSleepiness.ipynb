```python
import pandas as pd
import numpy as np
```

```python
sleep=pd.read_csv("sleep_deprivation_dataset_detailed.csv")
```

```python
sleep
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(style="whitegrid")
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=sleep, x='Caffeine_Intake', y='Daytime_Sleepiness')
plt.title("Caffeine vs Daytime Sleepiness")
plt.xlabel("Caffeine portions in a day")
plt.ylabel("Level of a daytime sleepiness")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=sleep, x='Caffeine_Intake', y='PVT_Reaction_Time')
plt.title("Caffeine vs Reactions time (PVT)")
plt.xlabel("Caffeine portions in a day")
plt.ylabel("Reactions time (ms)")
plt.show()
```

```python
correlation_sleep = sleep['Caffeine_Intake'].corr(sleep['Daytime_Sleepiness'])
correlation_reaction = sleep['Caffeine_Intake'].corr(sleep['PVT_Reaction_Time'])

print(f"Сorrelation between caffeine and sleepiness: {correlation_sleep:.2f}")
print(f"Correlation between caffeine and reaction time: {correlation_reaction:.2f}")
```

```python
plt.figure(figsize=(8, 5))
sns.regplot(data=sleep, x='Caffeine_Intake', y='Daytime_Sleepiness', ci=None, scatter_kws={'alpha':0.5})
plt.title("The Relations Between Caffeine and Daytime Sleepiness")
plt.xlabel("Caffeine portions")
plt.ylabel("Daytime Sleepiness")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.regplot(data=sleep, x='Caffeine_Intake', y='PVT_Reaction_Time', ci=None, scatter_kws={'alpha':0.5})
plt.title("Caffeine and Reactions time")
plt.xlabel("Caffeins portions")
plt.ylabel("Reactions time (ms)")
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.boxplot(data=sleep, x='Caffeine_Intake', y='Daytime_Sleepiness', palette='coolwarm', hue='Caffeine_Intake', legend=False)
plt.title("Sleepiness depending on caffeine intakes")
plt.xlabel("Caffeine portions a day")
plt.ylabel("Daytime sleepiness")
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.barplot(data=sleep, x='Caffeine_Intake', y='Daytime_Sleepiness', estimator='mean', errorbar='sd', palette='viridis', hue='Caffeine_Intake', legend=False)
plt.title("Average sleepiness based on caffeine intakes")
plt.xlabel("Caffeine portions")
plt.ylabel("Average sleepiness")
plt.show()
```

```python
plt.figure(figsize=(10, 8))
correlation_matrix = sleep.corr(numeric_only=True)
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Relations between columns")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.lmplot(data=sleep, x='Caffeine_Intake', y='PVT_Reaction_Time', hue='Gender', height=5, aspect=1.4)
plt.title("Caffeine and reaction time by gender")
plt.xlabel("Caffeine intakes")
plt.ylabel("Reaction time (ms)")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Загружаем данные
df = pd.read_csv("sleep_deprivation_dataset_detailed.csv")

# Строим график
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='Caffeine_Intake', y='Daytime_Sleepiness', errorbar='sd', palette='Blues', hue='Caffeine_Intake', legend=False)

plt.title("Sleepiness and caffeine")
plt.xlabel("Caffeine portions a day")
plt.ylabel("Average daytime sleepiness")
plt.ylim(0, df["Daytime_Sleepiness"].max() + 1)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
